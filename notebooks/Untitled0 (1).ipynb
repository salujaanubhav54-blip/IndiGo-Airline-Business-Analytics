```python
# IndiGo Airline Business Analytics

## Project Objective
The objective of this project is to analyze IndiGo Airlines’ operational and financial
performance using business analytics techniques and derive actionable insights to
improve route profitability and operational efficiency.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("IndiGo_Airline_Operations_Data.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
route_profit = df.groupby("Route")["Operating_Profit_INR"].sum().sort_values(ascending=False)

route_profit.plot(kind="bar", figsize=(10,5))
plt.title("Route-wise Operating Profit")
plt.show()
```

```python
# IndiGo Airline Business Analytics

## Business Context
IndiGo is India’s largest low-cost airline, operating at very high scale. While high
passenger demand drives revenue, operational inefficiencies such as delays and cost
overruns can negatively impact profitability. This project applies business analytics
to evaluate route performance and operational efficiency.
```

```python
## Exploratory Data Analysis (EDA)
This section explores the distribution of key operational and financial metrics to
understand route-level performance and identify potential inefficiencies.
```

```python
delay_profit = df.groupby("Avg_Delay_Min")["Operating_Profit_INR"].mean()

delay_profit.plot(kind="line", figsize=(10,5))
plt.xlabel("Average Delay (Minutes)")
plt.ylabel("Average Operating Profit (INR)")
plt.title("Impact of Delays on Profitability")
plt.show()
```

```python
## Key Business Insights
- Routes with higher delays show lower average operating profits.
- High passenger demand alone does not ensure profitability.
- Operational efficiency plays a critical role in airline margins.
- Certain routes deliver stable margins despite lower volumes.
```

```python
## Managerial Recommendations
- Focus on reducing delays on high-traffic routes to protect margins.
- Evaluate routes using profit-based KPIs rather than load factor alone.
- Improve cost efficiency through better scheduling and turnaround planning.
- Use analytics-driven insights for network and capacity planning decisions.
```
