```python
# IndiGo Airline Business Analytics

## Project Objective
The objective of this project is to analyze IndiGo Airlines’ operational and financial
performance using business analytics techniques and derive actionable insights to
improve route profitability and operational efficiency.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("IndiGo_Airline_Operations_Data.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
route_profit = df.groupby("Route")["Operating_Profit_INR"].sum().sort_values(ascending=False)

route_profit.plot(kind="bar", figsize=(10,5))
plt.title("Route-wise Operating Profit")
plt.show()
```
